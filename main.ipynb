```python
import pandas as pd

# Carrega os dados
df_janeiro = pd.read_csv('datasets/CATs/D.SDA.PDA.005.CAT.202601.csv', encoding='latin-1', sep=';')
df_fevereiro = pd.read_csv('datasets/CATs/D.SDA.PDA.005.CAT.202602.csv', encoding='latin-1', sep=';')
df_marco = pd.read_csv('datasets/CATs/D.SDA.PDA.005.CAT.202603.csv', encoding='latin-1', sep=';')
df_abril = pd.read_csv('datasets/CATs/D.SDA.PDA.005.CAT.202604.csv', encoding='latin-1', sep=';')
df_junho = pd.read_csv('datasets/CATs/D.SDA.PDA.005.CAT.202606.csv', encoding='latin-1', sep=';')

# Remove espaços em branco invisíveis de todas as colunas de texto ao carregar
for df_temp in [df_janeiro, df_fevereiro, df_marco, df_abril, df_junho]:
    df_temp.columns = df_temp.columns.str.replace(r'\s+', ' ', regex=True).str.strip()
    for col in df_temp.select_dtypes(include=['object']).columns:
        df_temp[col] = df_temp[col].astype(str).str.strip()
```

```python
df_semestre = pd.concat([df_janeiro, df_fevereiro, df_marco, df_abril, df_junho], ignore_index=True)
df_semestre.columns = df_semestre.columns.str.replace(r'\s+', ' ', regex=True).str.strip()

df_semestre.head()
```

```python
restricoes = {
    'UF Munic. Acidente': '{ñ class}',
    'UF Munic. Empregador': 'Ceará',
    'Munic Empr': '230440-Fortaleza',
}
```

```python
df_semestre_fortaleza = df_semestre.copy()

for restricao, alvo in restricoes.items():
    df_semestre_fortaleza = df_semestre_fortaleza[df_semestre_fortaleza[restricao] == alvo]

df_semestre_fortaleza.head()
```
